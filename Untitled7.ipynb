```python
import re
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations
from tokenizers import Tokenizer, models, trainers, pre_tokenizers
from gensim.models import Word2Vec
from sklearn.decomposition import PCA

np.random.seed(0)


pula_budowla = [
    "Stary zamek stoi na wysokim wzgórzu nad rzeką.",
    "Turyści zwiedzają zabytkowy zamek z przewodnikiem.",
    "Król zbudował potężny zamek z grubymi murami.",
    "Średniowieczny zamek ma wysoką kamienną wieżę.",
    "Wokół fosy ten zamek otacza głęboka woda.",
    "Zniszczony zamek przeszedł gruntowną renowację murów.",
    "Z wieży widać całą okolicę i piękny zamek w dolinie.",
    "Rycerze bronili dzielnie swojego miejsca jakim był zamek królewski.",
    "Ten zamek przyciąga latem wielu turystów z całego kraju.",
    "Kamienny zamek kryje w podziemiach dawne lochy.",
    "Królowa mieszkała przez lata w miejscu znanym jako zamek na skale.",
    "Zabytkowy zamek wpisano na listę cennych pomników historii.",
    "Mury obronne chroniły zamek przed atakiem wroga.",
    "Na dziedzińcu widać stary zamek i kamienną studnię.",
    "Zwiedzający podziwiają królewski zamek oraz bogate komnaty.",
    "Legenda głosi że ten zamek zamieszkuje biała dama.",
    "Wysoka wieża dominuje ponad budowlą jaką jest zamek obronny.",
    "Dawny zamek pełni obecnie funkcję muzeum regionalnego.",
    "Masywne mury otaczają zamek położony nad samym jeziorem.",
    "Architekt odnowił zabytkowy zamek zachowując dawny styl."
]

pula_mechanizm = [
    "Ślusarz naprawił zepsuty zamek w ciężkich drzwiach.",
    "Klucz nie pasuje bo zamek zaciął się rano.",
    "Nowy zamek antywłamaniowy chroni mieszkanie przed złodziejem.",
    "Wymieniłem stary zamek na solidny model z atestem.",
    "Zacięty zamek wymaga nasmarowania specjalnym olejem.",
    "Wkładka i zamek drzwiowy zostały szybko zamontowane.",
    "Złodziej próbował otworzyć zamek za pomocą wytrycha.",
    "Solidny zamek zatrzasnął się gdy wyszedłem z domu.",
    "Klamka i zamek w drzwiach wejściowych działają płynnie.",
    "Ze względu na awarię ślusarz rozwiercił stary zamek w bramie.",
    "Ten zamek posiada podwójny rygiel oraz mocny klucz.",
    "Kupiłem w sklepie metalowym nowy zamek do furtki.",
    "Zepsuty zamek uniemożliwił otwarcie drzwi do garażu.",
    "Precyzyjny mechanizm jaki ma zamek wymaga dopasowanego klucza.",
    "Zamek magnetyczny działa bardzo cicho w nowoczesnych drzwiach.",
    "Fachowiec szybko zamontował nowy zamek w stalowej bramie.",
    "Zamek zatrzaskowy blokuje drzwi natychmiast po ich zamknięciu.",
    "Awaria sprawiła że zamek przestał reagować na przekręcenie klucza.",
    "Mocny zamek stanowi podstawowe zabezpieczenie każdych drzwi.",
    "Ślusarz dorobił klucz i wyregulował zamek w biurze."
]

pula_neutralna = [
    "Turyści chętnie odwiedzają stare muzeum w centrum miasta.",
    "Zabytkowe mury i wysoka wieża przyciągają wielu zwiedzających.",
    "Król oraz królowa rządzili krajem przez długie lata.",
    "Rycerze nosili ciężkie zbroje podczas dawnych bitew.",
    "Przewodnik opowiada ciekawą historię o dawnych czasach.",
    "Ślusarz posiada w warsztacie różne narzędzia oraz klucze.",
    "Ciężkie drzwi wejściowe wykonano z litego drewna dębowego.",
    "Klucz leży na drewnianym stole obok klamki.",
    "Bezpieczeństwo mieszkania zależy od solidnych drzwi.",
    "Fachowiec naprawił klamkę oraz zawiasy w stalowej bramie.",
    "Pogoda dzisiaj sprzyja długim spacerom po parku.",
    "Książka historyczna opisuje życie ludzi w średniowieczu.",
    "Samochód jedzie szybko po szerokiej drodze asfaltowej.",
    "Komputer przetwarza duże zbiory danych tekstowych.",
    "Student przygotowuje sprawozdanie z ćwiczeń laboratoryjnych.",
    "Kawa poranna dodaje energii przed ciężką pracą.",
    "Pies biega wesoło po zielonym ogrodzie za domem.",
    "Kot śpi spokojnie na miękkim fotelu w salonie.",
    "Deszcz pada od rana i tworzy kałuże na ulicy.",
    "Muzyka klasyczna pomaga skupić się podczas nauki."
]


def normalizuj(zdania):
    return [re.sub(r'[^\w\s]', '', s.lower()).strip() for s in zdania]

norm_budowla = normalizuj(pula_budowla)
norm_mechanizm = normalizuj(pula_mechanizm)
norm_neutralna = normalizuj(pula_neutralna)
caly_korpus = norm_budowla + norm_mechanizm + norm_neutralna

print(f"Liczebność pul: budowla={len(norm_budowla)}, mechanizm={len(norm_mechanizm)}, neutralna={len(norm_neutralna)} | Razem={len(caly_korpus)}")

assert all(s.split().count("zamek") == 1 for s in norm_budowla), "Błąd w puli budowla!"
assert all(s.split().count("zamek") == 1 for s in norm_mechanizm), "Błąd w puli mechanizm!"
assert all(s.split().count("zamek") == 0 for s in norm_neutralna), "Błąd w puli neutralna!"
print("Weryfikacja liczby wystąpień słowa 'zamek' w pulach: OK!")

slowa_w_korpusie = set(w for s in caly_korpus for w in s.split())

# Zmodyfikowano pary na takie, w których druga forma (niewidziana) rzeczywiście nie występuje w korpusie
pary_bpe = [
    ("zamek", "zamkom"),
    ("drzwiach", "drzwiom"),
    ("mury", "murom"),
    ("klucz", "kluczom"),
    ("turyści", "turystom"),
    ("ślusarz", "ślusarzom")
]

formy_widziane = [p[0].lower() for p in pary_bpe]
formy_niewidziane = [p[1].lower() for p in pary_bpe]

assert all(w in slowa_w_korpusie for w in formy_widziane), "Któraś forma widziana nie jest w korpusie!"
assert all(w not in slowa_w_korpusie for w in formy_niewidziane), "Któraś forma niewidziana jest w korpusie!"
print("Weryfikacja 6 par form BPE (widziane vs niewidziane): OK!")

limity_bpe = [120, 250, 500]
tokenizery = {}

print("\n--- WYNIKI BPE ---")
for lim in limity_bpe:
    tok = Tokenizer(models.BPE(unk_token='[UNK]'))
    tok.pre_tokenizer = pre_tokenizers.Whitespace()
    trainer = trainers.BpeTrainer(vocab_size=lim, special_tokens=['[UNK]'], min_frequency=1)
    tok.train_from_iterator(caly_korpus, trainer)
    tokenizery[lim] = tok

    real_size = tok.get_vocab_size()


    enc_widz = [tok.encode(w).tokens for w in formy_widziane]
    avg_widz = np.mean([len(t) for t in enc_widz])
    unk_widz = sum(1 for t in enc_widz if '[UNK]' in t)

    enc_niewidz = [tok.encode(w).tokens for w in formy_niewidziane]
    avg_niewidz = np.mean([len(t) for t in enc_niewidz])
    unk_niewidz = sum(1 for t in enc_niewidz if '[UNK]' in t)

    print(f"Limit={lim:3d} | Faktyczny |V|={real_size:3d} | Widziane: śr={avg_widz:.2f}, UNK={unk_widz} | Niewidziane: śr={avg_niewidz:.2f}, UNK={unk_niewidz}")

print("\nPodziały 2 wybranych niewidzianych form ('zamkom', 'ślusarzom') dla wszystkich limitów:")
for forma in ["zamkom", "ślusarzom"]:
    print(f"Forma: '{forma}'")
    for lim in limity_bpe:
        print(f"  limit={lim:3d} (|V|={tokenizery[lim].get_vocab_size():3d}) -> {tokenizery[lim].encode(forma).tokens}")



wariant_1 = norm_neutralna + norm_budowla[:16] + norm_mechanizm[:4]
wariant_2 = norm_neutralna + norm_budowla[:4] + norm_mechanizm[:16]

zdania_w1 = [s.split() for s in (wariant_1 * 10)]
zdania_w2 = [s.split() for s in (wariant_2 * 10)]

w2v_1 = Word2Vec(zdania_w1, vector_size=50, window=4, min_count=5, sg=1, epochs=100, seed=0, workers=1)
w2v_2 = Word2Vec(zdania_w2, vector_size=50, window=4, min_count=5, sg=1, epochs=100, seed=0, workers=1)

print("\n--- WYNIKI WORD2VEC: TOP 5 SĄSIADÓW SŁOWA 'zamek' ---")
top5_1 = w2v_1.wv.most_similar("zamek", topn=5)
top5_2 = w2v_2.wv.most_similar("zamek", topn=5)

print("Wariant 1 (16 budowla / 4 mechanizm):")
for rank, (w, sim) in enumerate(top5_1, 1):
    in_w2 = "TAK" if w in w2v_2.wv else "NIE"
    print(f"  {rank}. {w:15s} cos={sim:.4f} (obecne w Wariancie 2: {in_w2})")

print("\nWariant 2 (4 budowla / 16 mechanizm):")
for rank, (w, sim) in enumerate(top5_2, 1):
    in_w1 = "TAK" if w in w2v_1.wv else "NIE"
    print(f"  {rank}. {w:15s} cos={sim:.4f} (obecne w Wariancie 1: {in_w1})")


grupa_budowla = ["zabytkowy", "mury", "wieża", "król", "turyści"]
grupa_mechanizm = ["ślusarz", "drzwiach", "klucz", "stary", "klamkę"]


grupa_budowla = [w for w in grupa_budowla if w in w2v_1.wv]
grupa_mechanizm = [w for w in grupa_mechanizm if w in w2v_1.wv]
print(f"\nSłowa w PCA - Grupa 1 (Budowla): {grupa_budowla}")
print(f"Słowa w PCA - Grupa 2 (Mechanizm): {grupa_mechanizm}")


def mean_cos_within(model, words):
    sims = [model.wv.similarity(w1, w2) for w1, w2 in combinations(words, 2)]
    return np.mean(sims), sims

def mean_cos_between(model, words1, words2):
    sims = [model.wv.similarity(w1, w2) for w1 in words1 for w2 in words2]
    return np.mean(sims), sims

cos_g1, _ = mean_cos_within(w2v_1, grupa_budowla)
cos_g2, _ = mean_cos_within(w2v_1, grupa_mechanizm)
cos_between, _ = mean_cos_between(w2v_1, grupa_budowla, grupa_mechanizm)

print("\n--- ŚREDNIE PODOBIEŃSTWO COSINUSOWE (Model 1) ---")
print(f"Wewnątrz Grupy 1 (Budowla):   {cos_g1:.4f}")
print(f"Wewnątrz Grupy 2 (Mechanizm): {cos_g2:.4f}")
print(f"Między Grupą 1 a Grupą 2:     {cos_between:.4f}")

wszystkie_pca = grupa_budowla + grupa_mechanizm + ["zamek"]
wektory = np.stack([w2v_1.wv[w] for w in wszystkie_pca])
coords = PCA(n_components=2, random_state=0).fit_transform(wektory)

plt.figure(figsize=(9, 6))
for (x, y), word in zip(coords, wszystkie_pca):
    if word == "zamek":
        plt.scatter(x, y, color='gold', edgecolor='black', s=180, marker='*', zorder=5, label='Badany wyraz (zamek)')
        plt.annotate(word.upper(), (x, y), xytext=(7, 5), textcoords='offset points', fontsize=12, fontweight='bold', color='black')
    elif word in grupa_budowla:
        plt.scatter(x, y, color='#2b5c8f', s=70, label='Grupa 1: Budowla' if word == grupa_budowla[0] else "")
        plt.annotate(word, (x, y), xytext=(5, 4), textcoords='offset points', fontsize=11, color='#2b5c8f')
    else:
        plt.scatter(x, y, color='#e0533c', s=70, label='Grupa 2: Mechanizm' if word == grupa_mechanizm[0] else "")
        plt.annotate(word, (x, y), xytext=(5, 4), textcoords='offset points', fontsize=11, color='#e0533c')

plt.title('Rzut PCA 2D embeddingów Word2Vec (Model 1: dominacja znaczenia "budowla")')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('pca_zamek_model1.png', dpi=150)
plt.show()
```
